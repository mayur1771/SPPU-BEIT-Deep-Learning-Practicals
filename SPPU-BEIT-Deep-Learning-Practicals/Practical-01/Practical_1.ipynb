```python
# Install once if required (run in a separate Jupyter cell)
%pip install tensorflow keras numpy

import tensorflow as tf
import keras
import numpy as np

print("DEEP LEARNING PACKAGES")
print("----------------------")

print("\n1. TensorFlow")
print("Developed by Google. Supports tensors, automatic differentiation and CPU/GPU computation.")

print("\n2. Keras")
print("High-level deep learning API used to build, train and evaluate neural networks.")

print("\n3. Theano")
print("A Python library for numerical computation and symbolic mathematical expressions. It is discontinued.")

print("\n4. PyTorch")
print("Developed by Meta. Supports tensor computation, automatic differentiation and deep learning.")

print("\nPACKAGE COMPARISON")
print("------------------")
print("TensorFlow : Numerical computation and deep learning framework.")
print("Keras      : High-level API for building neural networks.")
print("Theano     : Symbolic mathematical computation; discontinued.")
print("PyTorch    : Deep learning framework with dynamic computation graphs.")

print("\nTensorFlow version:", tf.__version__)
print("Keras version:", keras.__version__)
print("Available devices:", [d.device_type for d in tf.config.list_physical_devices()])

# TensorFlow tensor operations
a = tf.constant([[1., 2.], [3., 4.]])
b = tf.constant([[5., 6.], [7., 8.]])

print("\nTensor addition:\n", (a + b).numpy())
print("Matrix multiplication:\n", tf.matmul(a, b).numpy())

# Automatic differentiation
x = tf.Variable(3.0)
with tf.GradientTape() as tape:
    y = x ** 2 + 2 * x
print("Gradient:", tape.gradient(y, x).numpy())

# Simple Keras model
np.random.seed(0)
X = np.random.rand(500, 4).astype("float32")
Y = (X.sum(axis=1) > 2).astype("float32")

model = keras.Sequential([
    keras.layers.Input((4,)),
    keras.layers.Dense(8, activation="relu"),
    keras.layers.Dense(1, activation="sigmoid")
])

model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
model.summary()
model.fit(X, Y, epochs=15, batch_size=32, verbose=0)

print("Evaluation [loss, accuracy]:", model.evaluate(X, Y, verbose=0))
print("Prediction:", model.predict(np.array([[0.9, 0.8, 0.7, 0.6]]), verbose=0))
```
